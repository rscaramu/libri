# Soluzioni — capitolo 2 (esercizi misurati nell'appendice C)

```python
import sys; sys.path.insert(0, "..")
from utils import fissa_seme
import numpy as np
def sigmoide(z): return 1 / (1 + np.exp(-z))
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], float)
def neurone(y, tasso=0.5, epoche=2000, stocastico=False):
    np.random.seed(42); w = np.random.randn(2) * 0.1; b = 0.0; storia = []
    for _ in range(epoche):
        if stocastico:
            for i in np.random.permutation(4):
                yh = sigmoide(X[i] @ w + b); d = 2 * (yh - y[i]) * yh * (1 - yh); w -= tasso * d * X[i]; b -= tasso * d
            storia.append(np.mean((sigmoide(X @ w + b) - y) ** 2))
        else:
            yh = sigmoide(X @ w + b); storia.append(np.mean((yh - y) ** 2)); d = 2 * (yh - y) / 4 * yh * (1 - yh); w -= tasso * (X.T @ d); b -= tasso * d.sum()
    return w, b, storia
```

## Esercizio 1 — OR e NAND

```python
for nome, y in (("OR", np.array([0, 1, 1, 1.])), ("NAND", np.array([1, 1, 1, 0.]))):
    w, b, st = neurone(y); print(nome, np.round(sigmoide(X @ w + b), 3), "pesi", np.round(w, 2), "bias", round(b, 2))
```

## Esercizio 2 — il tasso di apprendimento

```python
for tasso in (1, 2, 5, 10, 20):
    w, b, st = neurone(np.array([0, 0, 0, 1.]), tasso=tasso); print(f"tasso {tasso:2d}: perdita finale {st[-1]:.5f}, monotona dopo l'epoca 100: {bool(np.all(np.diff(st[100:]) <= 1e-9))}")
```

## Esercizio 3 — stocastico contro batch

```python
y_and = np.array([0, 0, 0, 1.])
for stoc in (False, True):
    w, b, st = neurone(y_and, stocastico=stoc); st = np.array(st); print("stocastico" if stoc else "batch     ", "epoche per scendere sotto 0.01:", int(np.argmax(st < 0.01)) if np.any(st < 0.01) else "mai")
```
