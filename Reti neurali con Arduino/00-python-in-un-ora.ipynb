```python
# variabili, liste, cicli, condizioni
n = 128
assi = ['x', 'y', 'z']
for a in assi:
    if a != 'z':
        print(a, 'orizzontale')
    else:
        print(a, 'verticale')
```

```python
# funzioni
def normalizza(v, scala=4.0):
    return v / scala
print(normalizza(2.0))
```

```python
# NumPy: array, forma, reshape, statistiche
import numpy as np
finestra = np.random.default_rng(0).standard_normal((128, 3))   # 128 campioni x 3 assi
print(finestra.shape, finestra.mean(0), finestra.std(0))
vettore = finestra.flatten()                                     # 384 valori: x0 y0 z0 x1 y1 z1 ...
print(vettore.shape, vettore[:6])
```

```python
# pandas: un CSV in una tabella
import pandas as pd
df = pd.DataFrame(finestra, columns=['x', 'y', 'z'])
df.to_csv('finestra.csv', index=False)
print(pd.read_csv('finestra.csv').head())
```

```python
# Keras in cinque righe: la rete del capitolo 11
from tensorflow import keras
modello = keras.Sequential([keras.layers.Input((384,)), keras.layers.Dense(32, activation='relu'), keras.layers.Dense(3, activation='softmax')])
modello.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
modello.summary()   # 12.387 parametri, come nel paragrafo 2.5
```
